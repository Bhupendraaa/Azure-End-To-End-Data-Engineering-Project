```python
# Check Unity Catalog external locations and storage credentials

print("1. CHECKING EXTERNAL LOCATIONS")
print("=" * 60)

try:
    locations = spark.sql("SHOW EXTERNAL LOCATIONS")
    display(locations)

    location_names = [
        row["name"] for row in locations.collect()
    ]

    print("\n2. CHECKING LOCATION DETAILS")
    print("=" * 60)

    for name in location_names:
        print(f"\nExternal Location: {name}")

        try:
            details = spark.sql(
                f"DESCRIBE EXTERNAL LOCATION `{name}`"
            )
            display(details)
        except Exception as e:
            print("Could not describe:", str(e)[:500])

except Exception as e:
    print("ERROR:", str(e)[:1000])
```

### # Sales

```python
df_sales = df_sales.withColumn('StockDate', to_timestamp('StockDate'))
```

```python
from pyspark.sql.functions import *

df_sales = df_sales.withColumn(
    'OrderNumber',
    regexp_replace(col('OrderNumber'), 'S', 'T')
)
```

```python
df_sales = df_sales.withColumn('multiply', col('orderlineitem') * col('orderquantity'))
```

```python
df_sales.display()
```
